```python
import pandas as pd
import numpy as np
# Таблица 1: Продажи
sales = pd.DataFrame({
'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
'Price': [50000, 30000, 20000, 50000, np.nan,
50000, 20000, 30000, 50000, np.nan],
'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})
# Таблица 2: Информация о товарах
products = pd.DataFrame({
'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})
print("Таблица продаж:")
print(sales)
print("\nТаблица товаров:")
print(products)
```

```python
1. Найти стандартное отклонение Price для каждого продукта.
```

```python
print(sales.groupby('Product')['Price'].std())
```

```python
2. Найти регион с наибольшим количеством заказов.
```

```python
print(sales['Region'].value_counts().idxmax())
```

```python
3. Заполнить пропуски в Price медианой столбца.
```

```python
median_price = sales['Price'].median()
print("Медиана Price:", median_price)

sales['Price'] = sales['Price'].fillna(median_price)

print("\nТаблица после заполнения пропусков:")
print(sales)
```

```python
4. Отсортировать таблицу сначала по Product , затем по Quantity (оба по возрастанию).
```

```python
sales = sales.sort_values(['Product', 'Quantity'])
print(sales)
```

```python
5. Объединить sales и products так, чтобы остались только товары из products (right join).
```

```python
result = sales.merge(products, on='Product', how='right')
print(result)
```
