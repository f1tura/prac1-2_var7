```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
    'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
    'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
    90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 1. Вывести названия всех столбцов
```

```python
# 3. Выбрать строки, где Age меньше среднего возраста
```

```python
print(df[df['Age'] < df['Age'].mean()])
```

```python
df['Rank'] = df['Salary'].rank(ascending=False)

print(df)
```
